# A/B Testing & Hypothesis Testing

## Step 1: Hypotheses

- **H0 (Null):** The conversion rate of Control (A) equals the conversion rate of Variant (B), so p_A = p_B.
- **H1 (Alternative):** The conversion rates are different, so p_A ≠ p_B (two-tailed test).
- **Significance level:** α = 0.05

```python
import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.proportion import proportions_ztest
import matplotlib.pyplot as plt

df = pd.read_csv("checkout_ab_test_v2.csv")
df.head()
```

```python
df['converted_num'] = df['Converted'].map({'Yes': 1, 'No': 0})
```

```python
summary = df.groupby('Variant')['converted_num'].agg(['count', 'sum', 'mean'])
summary
```

```python
conv_A = summary.loc['A', 'sum']
conv_B = summary.loc['B', 'sum']
n_A = summary.loc['A', 'count']
n_B = summary.loc['B', 'count']
```

```python
z_stat, p_value = proportions_ztest([conv_A, conv_B], [n_A, n_B])
print(z_stat, p_value)
```

```python
table = pd.crosstab(df['Variant'], df['Converted'])
chi2, p_chi, dof, expected = stats.chi2_contingency(table, correction=False)
print(chi2, p_chi)
```

**Result:** p = 0.19 > α = 0.05, so we fail to reject H0. The chi-square test (χ² = 1.71, p = 0.19) confirms the Z-test.

```python
# 4a. Conversion rates and difference
p_A = conv_A / n_A
p_B = conv_B / n_B
diff = p_B - p_A

# 4b. Standard error of the difference
se = np.sqrt(p_A * (1 - p_A) / n_A + p_B * (1 - p_B) / n_B)

# 4c. 95% confidence interval for (B - A)
z_crit = 1.96
lower = diff - z_crit * se
upper = diff + z_crit * se

# 4d. Effect size
abs_lift = diff * 100            # percentage points
rel_lift = (diff / p_A) * 100    # % improvement over A

# Cohen's h (optional bonus effect size)
cohens_h = 2 * np.arcsin(np.sqrt(p_B)) - 2 * np.arcsin(np.sqrt(p_A))

print(f"Conversion A: {p_A:.4f} ({p_A*100:.2f}%)")
print(f"Conversion B: {p_B:.4f} ({p_B*100:.2f}%)")
print(f"Difference (B - A): {diff:.4f}")
print(f"Standard error: {se:.4f}")
print(f"95% CI: [{lower:.4f}, {upper:.4f}]")
print(f"Absolute lift: {abs_lift:.2f} percentage points")
print(f"Relative lift: {rel_lift:.2f}%")
print(f"Cohen's h: {cohens_h:.3f}")
```

**Result:** The 95% CI [-1.2 pp, +5.8 pp] includes 0, which agrees with the p-value. The effect size is very small (Cohen's h = 0.076).

```python
from statsmodels.stats.power import NormalIndPower
n_needed = NormalIndPower().solve_power(effect_size=cohens_h, alpha=0.05, power=0.8)
print(n_needed)
```

```python
print("Duplicates:", df['VisitorID'].duplicated().sum())
print(pd.crosstab(df['CheckoutStarted'], df['Converted']))
```

**Data check:** No duplicate visitors, and no conversions without a checkout start, so the data is consistent.

```python
rates = [p_A*100, p_B*100]
errs = [1.96*np.sqrt(p_A*(1-p_A)/n_A)*100, 1.96*np.sqrt(p_B*(1-p_B)/n_B)*100]
plt.bar(['A (Control)', 'B (Variant)'], rates, yerr=errs, capsize=8, color=['#999', '#4a7bd6'])
plt.ylabel('Conversion rate (%)')
plt.title('Checkout Conversion: A vs B (95% CI)')
plt.show()
```

```python
pd.DataFrame({
    'Metric': ['Visitors per group', 'Rate A', 'Rate B', 'Abs lift (pp)', 'Rel lift (%)',
               'Z-test p-value', 'Chi-square p-value', '95% CI (B-A)', "Cohen's h", 'Needed per group'],
    'Value': [600, f"{p_A:.2%}", f"{p_B:.2%}", f"{abs_lift:.2f}", f"{rel_lift:.2f}",
              f"{p_value:.4f}", f"{p_chi:.4f}", f"[{lower:.4f}, {upper:.4f}]",
              f"{cohens_h:.3f}", int(n_needed)]
})
```

## Executive Decision Memo

**Recommendation:** Keep Control (do not launch Variant B yet) and continue the test.

**Key numbers:** A = 9.50%, B = 11.83%, lift = +2.33 pp (+24.6% relative), Z-test p = 0.19, 95% CI for the difference [-1.2 pp, +5.8 pp], Cohen's h = 0.076.

**Reasoning:** B converted better in this sample, but the p-value (0.19) is above 0.05 and the confidence interval includes 0, so we cannot rule out that the gap is just random noise. The data does not support launching B.

**Caveat / next step:** Only 600 visitors per group were tested; about 2,740 per group are needed to detect a lift of this size reliably. Recommend extending the test to that sample size before making a final call.
